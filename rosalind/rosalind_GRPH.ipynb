```python
id_dict = {}
with open('rosalind_grph (1).txt', 'r') as file:
    current_id = ""
    for line in file:
        line = line.strip()
        if not line:
            continue
        if line[0] ==">":
            current_id = line[1:]
            id_dict[current_id] = ""
        else:
            id_dict[current_id] += line
```

```python
#print(id_dict)
```

```python
list_of_id = list(id_dict.keys())
for first_id in list_of_id:
    for second_id in list_of_id:
        if first_id == second_id:
            continue
        seq1 = id_dict[first_id]
        seq2 = id_dict[second_id]
        
        if seq1[-3:] == seq2[:3]:
            print(first_id, second_id)
```
